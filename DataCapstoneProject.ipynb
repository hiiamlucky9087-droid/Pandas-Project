**how many countries lie in the Eastern Europe Region**

```python
# by creating a manual function
def eastern_europe_countries(regions):
    count = 0
    for i in regions:
        if i == 'Eastern Europe':
            count += 1
    return count
```

```python
res = eastern_europe_countries(df['region'])
```

```python
res
```

**how many countries have Republic in their full name**

```python
df['country_long']
```

```python
def extract_republic(country):
    check = False
    count_republics = 0
    if "republic" in country.lower():
            check = True
            count_republics += 1
    return count_republics
```

```python
republics = df['country_long'].apply(extract_republic)
```

```python
republics.sum()
```
