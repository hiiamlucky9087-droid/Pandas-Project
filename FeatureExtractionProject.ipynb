```python
def extract_episodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True:
            data = data + i
        if i == "(":
            check = True
```

```python
df["Episodes"] = df["Title"].apply(extract_episodes) 
```

```python
df["Episodes"] = df["Episodes"].str.replace(" eps", "")
df
```

```python
def extract_TimeStamp(txt):
    index = txt.find(")")
    # slicing is better than loop for extracting the time stamp
    return txt[index + 1 : index + 20].strip()
```

```python
result = df["Title"].apply(extract_TimeStamp)
```

```python
result
```

```python
df["TimeStamp"] = df["Title"].apply(extract_TimeStamp)
df
```

```python
def extract_HighestScore(df):
    best_row_idx = df['Score'].idxmax()
    return df.loc[best_row_idx]["Title"]

highest_Score_Show = extract_HighestScore(df)
highest_Score_Show
```

```python
# 2nd approach to extract time stamps
def extract_time(txt):
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1, i+20):
                data += txt[j]
            return data
```

```python
df["Times"] = df["Title"].apply(extract_time)
df.head()
```

**which is the longest running anime**

```python
# caculating number of months each anime runned
df['Months'] = df['TimeStamp'].apply(
    lambda x: (
        pd.Period(x.split(' - ')[1], freq='M') -
        pd.Period(x.split(' - ')[0], freq='M')
    ).n
)

df
```

```python
# extracting the longest running episode
longest_running_anime = df[df["Months"] == df["Months"].max()]["Title"]
longest_running_anime
```
